# Feature Engineering

```python
df["Age"] = 2026-df["Year_Birth"]
```

```python
ref_date = df["Dt_Customer"].max()

df["Customer_Tenure_days"] = (ref_date-df["Dt_Customer"]).dt.days
```

```python
df["Total_Spending"] = df["MntWines"] + df["MntFruits"] +df["MntMeatProducts"] + df["MntFishProducts"] + df["MntSweetProducts"] +   df["MntGoldProds"] 
```

```python
df ["Total_children"] = df["Teenhome"] +  df["Kidhome"]
```

```python
df["Marital_Status"].value_counts()
```

```python
df["Education"] = df["Education"].replace({
    "Basic": "Undergraduate",
    "2n Cycle": "Undergraduate",
    "Graduation": "Graduate",
    "Master": "Master",
    "PhD": "Master"
})
```

```python
print(df["Education"].value_counts())
```

```python
df["Living_With"] = df["Marital_Status"].replace({
    "Married": "Patner" , "Together": "Patner",
    "Single":"Alone" ,"Divorced":"Alone" , "Widow" :"Alone" ,
    "Alone":"Alone", "Absurd":"Alone" ,"YOLO":"Alone"
})
```

```python
df.rename(columns={"Education ": "Education"}, inplace=True)
```

```python
df["Education"].value_counts()
```

```python
df["Education"] = df["Education"].replace({
    "Basic": "Underg",
    "2n Cycle": "Underg",
    "Graduation": "Graduate",
    "Master": "Master",
    "PhD": "Master"
})
```

# Encoding

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder()

cat_cols = ["Education","Living_With"]

enc_cols = ohe.fit_transform(df_cleaned[cat_cols])
```

```python
enc_df = pd.DataFrame(
    enc_cols.toarray(),
    columns=ohe.get_feature_names_out(cat_cols),
    index=df_cleaned.index
)
```

```python
enc_df.head()
```

```python
df_encoded = pd.concat([df_cleaned.drop(columns=cat_cols),enc_df],axis=1)
```

```python
df_encoded.shape
```

# scaling 

```python
from sklearn.preprocessing import StandardScaler
```

```python
X = df_encoded
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

# Visualization

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components = 4)

x_pca = pca.fit_transform(X_scaled)
```

```python
fig=plt.figure(figsize=(8,6))

ax=fig.add_subplot(111,projection="3d")

ax.scatter(x_pca[:,0],x_pca[:,1],x_pca[:,2])

ax.set_xlabel("PCA1")
ax.set_ylabel("PCA1")
ax.set_zlabel("PCA1")
ax.set_title("3d projection")
```

```python
pca.explained_variance_ratio_
```

# K value analysis

# 1 elbow method

```python
from sklearn.cluster import KMeans
from kneed import KneeLocator
```

```python
WCSS = []
for k in range(1,11):
    kmeans = KMeans(n_clusters=k,random_state=42)
    kmeans.fit_predict(x_pca)
    WCSS.append(kmeans.inertia_)
    
```

```python
kne = KneeLocator(range(1,11),WCSS,curve="convex",direction="decreasing")
```

```python
optimal_k = kne.elbow
```

```python
print("best k= ",optimal_k)
```

```python
plt.plot(range(1,11),WCSS,marker='o')
```

# sihoutte score 

```python
from sklearn.metrics import silhouette_score
```

```python
scores=[ ]
for k in range(2,11):
    kmeans = KMeans(n_clusters=k,random_state=42)
    labels = kmeans.fit_predict(x_pca)
    score = silhouette_score(x_pca,labels)
    scores.append(score)

plt.plot(range(2,11),scores,marker='o')
plt.xlabel("K")
plt.ylabel("Silhouette score")
```

```python
k_range = range(2,11)

fig, ax1 = plt.subplots(figsize=(10,6))
ax1.plot(k_range, WCSS, marker="o", color="blue")
```
